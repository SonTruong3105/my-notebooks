## 4.3 – Function (Hàm)

```python
# ===== CODE MẪU: Function =====
def calculate_average(numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

def find_max(numbers):
    if not numbers:
        return None
    max_val = numbers[0]
    for n in numbers:
        if n > max_val:
            max_val = n
    return max_val

# Gọi hàm
my_list = [1, 2, 3, 4, 5]
print("Trung bình:", calculate_average(my_list))
print("Trung bình List rỗng:", calculate_average([]))
print("Max:", find_max(my_list))
```

## 6.1 – Tư duy Vectorization

```python
# ===== CODE MẪU: So sánh tốc độ For loop vs NumPy =====
N = 1000000

# Cách 1: Vòng lặp for truyền thống
list_a = list(range(N))
start = time.perf_counter()
result_loop = [x + 5 for x in list_a]
time_loop = time.perf_counter() - start

# Cách 2: NumPy Vectorization
arr_a = np.arange(N)
start = time.perf_counter()
result_vec = arr_a + 5
time_vec = time.perf_counter() - start

print(f"For loop: {time_loop:.4f} giây")
print(f"NumPy:    {time_vec:.4f} giây")
print(f"NumPy nhanh hơn: {time_loop / time_vec:.1f} lần")
```

## 6.7 – Dot Product & Broadcasting

```python
# ===== CODE MẪU: Dot Product & Broadcasting =====
# --- Dot Product ---
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("A:\n", A)
print("B:\n", B)
print("\nA @ B (nhân ma trận):\n", A @ B)
print("np.dot(A, B):\n", np.dot(A, B))
print("\nA * B (nhân từng phần tử):\n", A * B)

# --- Broadcasting ---
X = np.array([[1, 2, 3],
              [4, 5, 6]])
bias = np.array([10, 20, 30])
scalar = 100

print("\n--- Broadcasting ---")
print("X + bias:\n", X + bias)
print("\nX + scalar:\n", X + scalar)
print("\nX * 2:\n", X * 2)
```

## 7.2 – Xử lý dữ liệu thiếu (NaN)

```python
# ===== CODE MẪU: Xử lý NaN =====
df = pd.DataFrame({
    'Name': ['An', 'Binh', 'Chi', 'Dung', 'Em'],
    'Age': [22, 38, 26, np.nan, 35],
    'City': ['Hanoi', 'HCM', 'Hanoi', 'Danang', 'HCM'],
    'Salary': [1000, 2000, 1500, 1200, 2500]
})

print("Số lượng NaN mỗi cột:")
print(df.isnull().sum())

# Điền NaN bằng giá trị trung bình
df['Age'] = df['Age'].fillna(df['Age'].mean())

print("\nSau khi điền NaN:")
print(df)

print("\nKiểm tra lại NaN:")
print(df.isnull().sum())
```

## 7.3 – Lọc dữ liệu

```python
# ===== CODE MẪU: Lọc dữ liệu =====
df = pd.DataFrame({
    'Name': ['An', 'Binh', 'Chi', 'Dung', 'Em'],
    'Age': [22, 38, 26, 30, 35],
    'City': ['Hanoi', 'HCM', 'Hanoi', 'Danang', 'HCM'],
    'Salary': [1000, 2000, 1500, 1200, 2500]
})

# Lọc 1 điều kiện
print("Người có Salary > 1500:")
print(df[df['Salary'] > 1500])

# Lọc nhiều điều kiện (dùng & và |)
print("\nNgười ở HCM VÀ Salary > 1500:")
print(df[(df['City'] == 'HCM') & (df['Salary'] > 1500)])

# Lọc theo danh sách
print("\nNgười ở Hanoi hoặc HCM:")
print(df[df['City'].isin(['Hanoi', 'HCM'])])
```

## 7.4 – Groupby & Tổng hợp

```python
# ===== CODE MẪU: Groupby =====
df = pd.DataFrame({
    'Name': ['An', 'Binh', 'Chi', 'Dung', 'Em'],
    'City': ['Hanoi', 'HCM', 'Hanoi', 'Danang', 'HCM'],
    'Salary': [1000, 2000, 1500, 1200, 2500]
})

# Groupby 1 cột, tính mean
print("Lương trung bình theo City:")
print(df.groupby('City')['Salary'].mean())

# Groupby với nhiều hàm tổng hợp
print("\nThống kê lương theo City:")
print(df.groupby('City')['Salary'].agg(['mean', 'max', 'min', 'count']))
```

## 7.5 – Sắp xếp dữ liệu

```python
# ===== CODE MẪU: Sort =====
df = pd.DataFrame({
    'Name': ['An', 'Binh', 'Chi', 'Dung', 'Em'],
    'Age': [22, 38, 26, 30, 35],
    'Salary': [1000, 2000, 1500, 1200, 2500]
})

print("Sắp xếp theo Salary giảm dần:")
print(df.sort_values(by='Salary', ascending=False))

print("\nSắp xếp theo Age tăng dần:")
print(df.sort_values(by='Age', ascending=True))

# Sắp xếp theo nhiều cột
df['City'] = ['Hanoi', 'HCM', 'Hanoi', 'Danang', 'HCM']
print("\nSắp xếp theo City (A-Z) và Salary (giảm):")
print(df.sort_values(by=['City', 'Salary'], ascending=[True, False]))
```
